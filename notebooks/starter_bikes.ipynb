```
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler

import pandas as pd
bikes = pd.read_csv('https://raw.githubusercontent.com/byui-cse/cse450-course/master/data/bikes.csv')

# import pandas as pl
# bikes = pl.read_csv('https://raw.githubusercontent.com/byui-cse/cse450-course/master/data/bikes.csv')
```

```
# Separate the features and define the target variable
x = bikes.drop(columns=['dteday', 'casual', 'registered'])
y = bikes['casual'] + bikes['registered']

# Split the data
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

# Fit the training data
scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```
# Build of the Neural Network
model = keras.Sequential([
    # Input later
    layers.Input(shape=(X_train_scaled.shape[1],)),

    # First hidden layer (64 Neurons)
    layers.Dense(64, activation='relu'),

    # Second hidden layer (32 Neurons)
    layers.Dense(32, activation='relu'),

    # Output layer
    layers.Dense(1)
])
```

```
model.compile(optimizer='adam', loss='mean_squared_error', metrics=['mean_absolute_error'])
```

```
# Train the model
history = model.fit(X_train_scaled, y_train, validation_split=0.2, epochs=50, batch_size=32)
```

```
# Evaluate model performance
results = model.evaluate(X_test_scaled, y_test)
print(print(f"Test Loss / MAE: {results}"))
```
